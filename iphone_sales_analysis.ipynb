# Question 1: What are the Top 10 Highest-Rated iPhones on Flipkart?

```python
top_10_rated = df.sort_values(by='Star Rating',ascending=False)[['Product Name','Star Rating']].head(10)

print(top_10_rated)
```

```python
plt.figure(figsize=(8,4))
sns.barplot(
    data=top_10_rated,
    x='Star Rating',
    y='Product Name'
)

plt.title('Top 10 Highest Rated iPhones')
plt.xlabel('Star Rating')
plt.ylabel('iPhone Model')
plt.show()
```

# Question 2:
# How Many Ratings Do the Highest-Rated iPhones Have?

```python
# Find highest star rating
highest_rating = df['Star Rating'].max()

# Get iPhones with highest star rating
highest_rated_iphones = df[df['Star Rating'] == highest_rating]

# Show product name, star rating, and number of ratings
print(highest_rated_iphones[['Product Name', 'Star Rating', 'Number Of Ratings']])


Another way : 

# highest_rated = df[df['star rating'] == df['star rating'].max()]

# result = highest_rated[[
#     'product name',
#     'star rating',
#     'number of ratings'
# ]]

# print(result)
```

###### Simple Explanation
#### max() finds the highest star rating in the dataset.

#### We filter the dataset to keep only iPhones that have that highest rating.

#### We display:
#### Product Name
#### Star Rating
#### Number of Ratings

#### Example Output
#### Product Name	Star Rating	Number of Ratings
#### iPhone 13 Pro Max	4.8	12,500
#### iPhone 14 Pro	4.8	9,800
#### Business Insight

#### The highest-rated iPhones are the models that customers are most satisfied with. 
#### The number of ratings shows how many customers have rated those products, helping us understand both customer satisfaction and product popularity.

```python
plt.figure(figsize=(6,6))

sns.barplot(
    data=highest_rated_iphones,
    x='Product Name',
    y='Number Of Ratings'
)

plt.xticks(rotation=90)
plt.title('Ratings of Highest Rated iPhones')
plt.show()
```

# Question 4:
# Relationship Between Sale Price and Number of Ratings

```python
correlation = df['Sale Price'].corr(df['Number Of Ratings'])

print("correlation:",correlation)
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x='Sale Price',
    y='Number Of Ratings'
)

plt.title('Sale Price vs Number of Ratings')
plt.xlabel('Sale Price')
plt.ylabel('Number of Ratings')
plt.show()
```
